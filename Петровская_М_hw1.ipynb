## Список фильмов

У Вас есть информация о названии, годе выхода, рейтинге и жанре художественных фильмов.

````python
movies = [
    ['Побег из Шоушенка', 1994, 9.3, 'драма'],
    ['Крёстный отец', 1972, 9.2, 'криминал'],
    ['Тёмный рыцарь', 2008, 9.0, 'боевик'],
    ['Криминальное чтиво', 1994, 8.9, 'криминал'],
    ['Форрест Гамп', 1994, 8.8, 'драма']
]
````

Задание 1. Выведите название, год и рейтинг фильма строго в формате:
```
1. Побег из Шоушенка (1994) — рейтинг: 9.3
2. Крёстный отец (1972) — рейтинг: 9.2
3. Тёмный рыцарь (2008) — рейтинг: 9.0
```

````python
for index, movie in enumerate(movies, 1):
    print(f'{index}. {movie[0]} ({movie[1]}) — рейтинг: {movie[2]}')
````

Задание 2. Выведите:
- общее количество фильмов в коллекции,
- количество боевиков,
- средний рейтинг,
- название самого старого фильма,
- навзание фильма с самым высоким рейтингом.

Выведите указанную выше статистику с использованием f-строки.

````python
from statistics import mean
print(f'Всего фильмоы: {len(movies)}')
action = 0
for movie in movies:
    if movie[3] == 'боевик':
        action += 1
print(f'Количество боевиков: {action}')

ratings = [movie[2] for movie in movies]
print(f'Средний рейтинг: {mean(ratings)}')

oldest = movies[0]
for movie in movies:
    if movie[1] < oldest[1]:
        oldest = movie
print(f'Самый старый: {oldest[0]}')

best = movies[0]
for movie in movies:
    if movie[2] > best[2]:
        best = movie
print(f'C самым высоким рейтингом: {best[0]}')
````

Задание 3. Найдите все фильмы, название которых начинается на определённую букву (запросите букву у пользователя с клавиатуры). Дайте пользователю возможность ввести букву в любом регистре. Если фильмов, начинающихся с такой буквы, нет, выведите сообщение «Фильмы не найдены».

````python
letter = input('Напечатайте букву: ').lower()
found = False
for movie in movies:
    if movie[0].lower().startswith(letter):
        print(movie[0])
        found = True
if not found:
    print('Фильмы не найдены')
````

Задание 4. Запросите у пользователя год. Найдите все фильмы, выпущенные в указанный год или позднее.
