У туристического агентства появился клиент, который хочет потратить на путешествие не больше определённой суммы.

Напишите функцию is_affordable(). Она должна возвращать True, если стоимость маршрута не превышает max_price; False в противном случае.

Затем напишите функцию get_affordable_routes(routes, max_price). Она должна:

- использовать filter() и функцию is_affordable;
- возвращать список маршрутов, подходящих по цене.

````python
def is_affordable(route, max_price):
    return route[3] <= max_price
def get_affordable_routes(routes, max_price):
    def affordable(route):
        return is_affordable(route, max_price)
    return list(filter(affordable, routes))
````

8. Нагрузка экскурсоводов.

Смешарики хотят распределить работу между экскурсоводами. Напишите функцию count_guide_routes(), которая возвращает количество маршрутов, которые проводит указанный экскурсовод. Затем напишите функцию get_guide_statistics(), которая принимает список экскурсоводов и для каждого экскурсовода определяет количество закреплённых за ним маршрутов.

````python
def count_guide_routes(routes, guide_name):
    count = 0
    for route in routes:
        if route[4] == guide_name:
            count += 1
    return count
def get_guide_statistics(routes, guides_list):
    stat = {}
    for guide in guides_list:
        stat[guide] = count_guide_routes(routes, guide)
    return stat
````


Необходимо написать функцию assign_guides(routes, guides), которая распределит экскурсоводов по маршрутам.

Для каждого маршрута нужно найти экскурсовода, который:

Программа должна работать с данными из списка guides.

Если для какого-либо маршрута подходящего экскурсовода не нашлось, вместо имени экскурсовода запишите: "Экскурсовод не найден".

````python
def assign_guides(routes, guides):
    counts = {}
    for guide in guides:
        counts[guide[0]] = 0
    result = []
    for route in routes:
        found = None
        for guide in guides:
            name = guide[0]
            directions = guide[1]
            max_routes = guide[2]
            if route[1] in directions and counts[name] < max_routes:
                found = name
                counts[name] += 1
                break
        if found:
            result.append(route + [found])
        else:
            result.append(route + ["Экскурсовод не найден"])
    return result
````
