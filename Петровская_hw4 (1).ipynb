````python
import os
import regex as re
import pandas as pd

from razdel import tokenize
import pymorphy3

morph = pymorphy3.MorphAnalyzer()
````

````python
source_text = source_text.replace("\r\n", "\n").replace("\r", "\n")

source_text = re.sub(r"\[\d+\]", "", source_text)
source_text = re.sub(r"[ \t]+\n", "\n", source_text)

print(source_text[:1500])
````

````python
letter_blocks = re.findall(
    r"\*\s*\*\s*\*(.+?)(?=\*\s*\*\s*\*|\[Воспоминания А\. И\. Степановой\]|$)",
    source_text,
    flags=re.DOTALL
)

letter_blocks = [block.strip() for block in letter_blocks if block.strip()]

print("Извлечено блок:", len(letter_blocks))
````

````python
letter_blocks = [
    re.sub(
        r"Они встретились в квартире Бориса Робертовича в 1957 году.*$",
        "",
        block,
        flags=re.DOTALL
    ).strip()
    for block in letter_blocks
]

letters = []
telegram_blocks = []

for block in letter_blocks:
    first_nonempty_line = next(
        (line.strip() for line in block.splitlines() if line.strip()),
        ""
    )

    if re.match(r"^Телеграмма\b", first_nonempty_line, flags=re.IGNORECASE):
        telegram_blocks.append(block)
    else:
        letters.append(block)

print("Осталось писем:", len(letters))
print("Исключено:", len(telegram_blocks))

if telegram_blocks:
    print("\nПроверка:")
    print(telegram_blocks[0][:1000])
````

````python
letters_df = pd.DataFrame({
    "letter_id": range(len(letters)),
    "raw_letter": letters
})

display(letters_df.head(5))
print("Всего записей:", len(letters_df))
````

````python
MONTHS = {
    "января": 1, "янв.": 1,
    "февраля": 2, "февр.": 2, "фев.": 2,
    "марта": 3, "мар.": 3,
    "апреля": 4, "апр.": 4,
    "мая": 5,
    "июня": 6, "июн.": 6,
    "июля": 7, "июл.": 7,
    "августа": 8, "авг.": 8,
    "сентября": 9, "сент.": 9, "сен.": 9,
    "октября": 10, "окт.": 10,
    "ноября": 11, "нояб.": 11, "ноя.": 11,
    "декабря": 12, "дек.": 12
}

ROMAN_MONTHS = {
    "I": 1, "II": 2, "III": 3, "IV": 4,
    "V": 5, "VI": 6, "VII": 7, "VIII": 8,
    "IX": 9, "X": 10, "XI": 11, "XII": 12
}

def normalize_year(year_string):
    year = int(year_string)
    if year < 100:
        year += 1900
    return year

def make_iso_date(day, month, year):
    try:
        day = int(day)
        month = int(month)
        year = int(year)
        if year < 100:
            year += 1900
        return f"{year:04d}-{month:02d}-{day:02d}" if 1 <= month <= 12 and 1 <= day <= 31 else None
    except (TypeError, ValueError):
        return None

def parse_date(letter):
    
    header = "\n".join(letter.splitlines()[:15])

    
    match = re.search(
        r"(?<!\d)(\d{1,2})[./](\d{1,2})[./](\d{2,4})(?!\d)",
        header
    )
    if match:
        day, month, year = match.groups()
        return make_iso_date(day, month, year)

    match = re.search(
        r"(?<!\d)(\d{1,2})\s*/\s*(I{1,3}|IV|V?I{0,3}|IX|X{1,2}|XI|XII)\s*[-.]\s*(\d{2,4})(?!\d)",
        header,
        flags=re.IGNORECASE
    )
    if match:
        day, roman_month, year = match.groups()
        month = ROMAN_MONTHS.get(roman_month.upper())
        if month:
            return make_iso_date(day, month, year)

    month_pattern = "|".join(
        sorted((re.escape(m) for m in MONTHS), key=len, reverse=True)
    )
    match = re.search(
        rf"(?<!\d)(\d{{1,2}})\s+({month_pattern})\s+(\d{{2,4}})(?:\s*(?:г\.?|года))?",
        header,
        flags=re.IGNORECASE
    )
    if match:
        day, month_name, year = match.groups()
        month = MONTHS.get(month_name.lower())
        if month:
            return make_iso_date(day, month, year)

    return None

letters_df["date"] = letters_df["raw_letter"].apply(parse_date)

print("Распределение дат:")
print("С датой:", letters_df["date"].notna().sum())
print("Без однозначнй даты:", letters_df["date"].isna().sum())

display(letters_df[["letter_id", "date"]].head(20))
````

````python
def detect_author(letter):
    lines = [line.strip() for line in letter.splitlines() if line.strip()]
    header = " ".join(lines[:12])
    ending = " ".join(lines[-8:])

    
    if re.search(
        r"А\.\s*О\.\s*Степановой|Ангелине Осиповне Степановой|"
        r"Степановой\s*[—-]\s*для\s*А\.\s*О\.\s*Степановой",
        header,
        flags=re.IGNORECASE
    ):
        return "Николай Эрдман"

    
    if re.search(r"(?:^|\s)(?:Лина|Ангелина|Л\.)[.!…]?$", ending):
        return "Ангелина Степанова"

    
    if re.search(r"(?:Николай|Н\.\s*Р\.\s*Эрдман|Н\.\s*Эрдман)[.!…]?$", ending):
        return "Николай Эрдман"

    return None

letters_df["author"] = letters_df["raw_letter"].apply(detect_author)

print("Автоматически опр авторы:")
print(letters_df["author"].value_counts(dropna=False))
````

````python
author_corrections = {
    9: "Николай Эрдман",
    21: "Николай Эрдман",
    25: "Николай Эрдман",
    26: "Николай Эрдман",
    32: "Николай Эрдман",
    33: "Ангелина Степанова",
    34: "Николай Эрдман",
    35: "Ангелина Степанова",
    38: "Николай Эрдман",
    62: "Николай Эрдман",
    69: "Николай Эрдман",
    71: "Николай Эрдман"
}

for letter_id, author in author_corrections.items():
    letters_df.loc[letters_df["letter_id"] == letter_id, "author"] = author

print("После коррект:")
print(letters_df["author"].value_counts(dropna=False))

print("\nПисьма без опр автора:")
display(letters_df[letters_df["author"].isna()][["letter_id", "raw_letter"]])
````

````python
def extract_body(letter):
    parts = re.split(r"\n\s*\n", letter.strip(), maxsplit=1)
    if len(parts) == 2 and parts[1].strip():
        return parts[1].strip()
    return letter.strip()

letters_df["text"] = letters_df["raw_letter"].apply(extract_body)

letters_df["text"] = letters_df["text"].apply(
    lambda text: re.sub(
        r"Они встретились в квартире Бориса Робертовича в 1957 году.*$",
        "",
        text,
        flags=re.DOTALL
    ).strip()
)

print("Примеры:")
for i in range(min(3, len(letters_df))):
    print("\nLETTER_ID:", letters_df.loc[i, "letter_id"])
    print(letters_df.loc[i, "text"][:700])
````

````python
check_ids = sorted(set([
    0,
    min(1, len(letters_df) - 1),
    len(letters_df) // 2,
    len(letters_df) - 2,
    len(letters_df) - 1
]))

for letter_id in check_ids:
    row = letters_df.loc[letters_df["letter_id"] == letter_id].iloc[0]
    print("\n" + "=" * 70)
    print("LETTER_ID:", letter_id, "| AUTHOR:", row["author"], "| DATE:", row["date"])
    print(row["text"][:1000])
````

````python
def clean_text(text):
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text)
    return text.strip()

letters_df["text"] = letters_df["text"].apply(clean_text)

display(letters_df[["letter_id", "author", "date", "text"]].head(5))
````

````python
def lemmatize(text):
    tokens = [token.text.lower() for token in tokenize(text)]
    lemmas = []

    for token in tokens:
        if token.isalnum():
            lemma = morph.parse(token)[0].normal_form
            lemmas.append(lemma)

    return " ".join(lemmas)

letters_df["lemmatized_text"] = letters_df["text"].apply(lemmatize)

display(
    letters_df[["text", "lemmatized_text"]].head(3)
)
````

````python
def count_words(text):
    tokens = [token.text for token in tokenize(text)]
    return sum(token.isalnum() for token in tokens)

letters_df["word_count"] = letters_df["text"].apply(count_words)

display(letters_df[["letter_id", "word_count"]].head(10))
````

````python
df = letters_df[
    ["date", "author", "text", "lemmatized_text", "word_count"]
].copy()

df = df[["date", "author", "text", "lemmatized_text", "word_count"]]

display(df.head())
print("Размер таблицы:", df.shape)
````
