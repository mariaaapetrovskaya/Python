### Задача 1
Напишите программу, которая считает:
- количество слов в тексте,
- количество уникальных слов,
- самое длинное слово,
- самое часто встречающееся слово.

Необходимо убрать знаки препинания и привести всё к нижнему регистру.

Ответ выведите в свободной форме с использованием f-строки, а также СОХРАНИТЕ его в текстовый файл "result.txt".

````python
from collections import Counter

text = """Жил-был у бабушки серенький козлик,
Жил-был у бабушки серенький козлик,
Вот как, вот как, серенький козлик,
Вот как, вот как, серенький козлик.

Бабушка козлика очень любила,
Бабушка козлика очень любила,
Вот как, вот как, очень любила,
Вот как, вот как, очень любила.

Вздумалось козлику в лес погуляти,
Вздумалось козлику в лес погуляти,
Вот как, вот как, в лес погуляти,
Вот как, вот как, в лес погуляти.

Напали на козлика серые волки,
Напали на козлика серые волки,
Вот как, вот, как серые волки,
Вот как, вот, как серые волки.

Остались от козлика рожки да ножки,
Остались от козлика рожки да ножки,
Вот как, вот как, рожки да ножки,
Вот как, вот как, рожки да ножки."""

sign = '.,!?;:()[]{}"\'`-—'

words = []
for token in text.split():
    word = token.strip(sign).lower()
    if word:
        words.append(word)

number_words = len(words)
unique_words = len(set(words))
long_word = max(words, key=len)
themost_word = Counter(words).most_common(1)[0][0]

result = (
    f"Количество слов в тексте: {number_words}\n"
    f"Количество уникальных слов: {unique_words}\n"
    f"Самое длинное слово: {long_word}\n"
    f"Самое часто встречающееся слово: {themost_word}"
)

print(result)

with open("result.txt", "w", encoding="utf-8") as file:
    file.write(result)
````

### Задача 2

Напишите программу, которая запрашивает у пользователя пароль и проверяет, что:
- длина пароля не меньше 8 символов;
- в пароле есть хотя бы одна цифра;
- в пароле есть хотя бы одна заглавная буква.

Цикл не прерывается до тех пор, пока не будет введён верный пароль. Если все условия выполнены — вывести «Пароль надёжный». Если какое-то условие не выполнено, выведи сообщение о том, какое именно требование было нарушено.

Используйте методы .isdigit(), .isupper() и функцию len().

````python
while True:
    password = input("Введите пароль: ")
    lengh = len(password) >= 8
    is_digit = False
    is_upper = False

    for ch in password:
        if ch.isdigit():
            is_digit = True
        if ch.isupper():
            is_upper = True

    if lengh and is_digit and is_upper:
        print("Пароль надежный")
        break
    else:
        if not lengh:
            print("Пароль должен содержать не менее 8 символов. Введите пароль еще раз.")
        if not is_digit:
            print("Пароль должен содержать хотя бы одну цифру.Введите пароль еще раз.")
        if not is_upper:
            print("Пароль должен содержать хотя бы одну заглавную букву.Введите пароль еще раз.")
````

### Задача 3

Кот Барсик ходит по подъезду и мяукает под дверями соседей. Кто-то даёт ему еду, кто-то — нет. Нужно выяснить, кто из соседей самый щедрый, а кто ~~самый жадный~~ больше всего заботится о правильном питании Барсика.

````python
# Соседи, под чьими дверями мяукал Барсик (по порядку обхода)
visits = [
    "Ивановы", "Петровы", "Ивановы", "Сидоровы",
    "Петровы", "Ивановы", "Сидоровы", "Ивановы",
    "Петровы", "Сидоровы", "Ивановы", "Петровы"
]

# Что дали Барсику при каждом визите
results = [
    "еда", "прогнали", "еда", "погладили",
    "еда", "прогнали", "прогнали", "еда",
    "погладили", "еда", "еда", "прогнали"
]

fed = {}
chased = {}
petted = {}

for i in range(len(visits)):
    neighbour = visits[i]
    did = results[i]

    if did == "еда":
        if neighbour not in fed:
            fed[neighbour] = 0
        fed[neighbour] += 1
    elif did == "прогнали":
        if neighbour not in chased:
            chased[neighbour] = 0
        chased[neighbour] += 1
    elif did == "погладили":
        if neighbour not in petted:
            petted[neighbour] = 0
        petted[neighbour] += 1

most_generous = max(fed, key=fed.get)
most_greedy = max(chased, key=chased.get)

total_fed = sum(fed.values())
total_chased = sum(chased.values())
total_petted = sum(petted.values())

print("Покормили:", fed)
print("Прогнали:", chased)
print("Погладили:", petted)
print()
print(f"Самый щедрый: {most_generous} (покормил {fed[most_generous]} раз)")
print(f"Самый жадный: {most_greedy} (прогнал {chased[most_greedy]} раз)")
print()
print("Общий 'баланс Барсика':")
print(f"  Покормили: {total_fed} раз")
print(f"  Погладили: {total_petted} раз")
print(f"  Прогнали: {total_chased} раз")
````

Нужно написать программу, которая:

1. Пройдёт циклом for по индексам (не по элементам, потому что нужны оба списка одновременно).
2. Соберёт три отдельных словаря fed (сколько раз покормили), chased (сколько раз прогнали), petted (сколько раз погладили) вида:
```
{ "Ивановы": 3,
    "Петровы": 2,
    ...}
```
Подсказка: прежде чем увеличивать счётчик, проверьте, есть ли уже сосед в словаре. Если нет — создайте для него значение 0.

3. Найдёт самого щедрого.
4. Найдёт самого жадного.
5. Посчитает общий "баланс Барсика": сколько раз его покормили, сколько погладили, сколько прогнали.
